# Eval Episode Viewer — 逐幀檢視測試任務

選 episode + 拉 frame 進度條，看該幀的**各關節值 + 水果位置 + 雙臂姿態(俯視/側視)**，
以及該集**入籃成功 / 回 home** 與否。資料來自 eval 時設 `EEF_LOG=<csv>` 產生的逐幀 log。

**用法**：改下面 `CSV_PATH` 指到你要看的 eval log → **Restart Kernel & Run All** → 用下拉+滑桿看。
（需要 `ipywidgets`；沒有的話 `pip install ipywidgets`。）


```python
import pandas as pd, numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import ipywidgets as widgets

# ── 選 eval 逐幀 log（EEF_LOG 產生）──
CSV_PATH = Path.home() / 'openarm_isaaclab/outputs/eval_view_appleV2.csv'
# 這次任務的目標水果/籃（判入籃用）
TARGET_FRUIT, TARGET_BASKET = 'apple', 'red'

df = pd.read_csv(CSV_PATH)
df = df[df.state.isin(['ready', 'running'])].reset_index(drop=True)

# 欄位分類
JOINTS = [c for c in df.columns if c.startswith('openarm_') and 'joint' in c and not c.endswith(('_x', '_y', '_z'))]
def arm_links(arm):
    order = [f'openarm_{arm}_link{i}' for i in range(1, 8)] + [f'openarm_{arm}_left_finger', f'openarm_{arm}_right_finger']
    return [b for b in order if f'{b}_x' in df.columns]
RIGHT, LEFT = arm_links('right'), arm_links('left')
FRUITS = [f for f in ['apple', 'strawberry', 'banana'] if f'{f}_x' in df.columns]

# 幾何
RB = dict(x=(0.198, 0.349), y=(-0.115, -0.031)); YB = dict(x=(0.198, 0.349), y=(-0.221, -0.136))
PLAT = dict(x=(0.166, 0.568), y=(-0.300, 0.300), top=0.193)
BASK = {'red': RB, 'yellow': YB}

episodes = sorted(df.episode.unique())
def ep_status(ep):
    last = df[df.episode == ep].iloc[-1]
    b = BASK[TARGET_BASKET]
    fx, fy, fz = last[f'{TARGET_FRUIT}_x'], last[f'{TARGET_FRUIT}_y'], last[f'{TARGET_FRUIT}_z']
    inb = b['x'][0] <= fx <= b['x'][1] and b['y'][0] <= fy <= b['y'][1] and fz < 0.28
    arm = max(abs(last[j]) for j in JOINTS if 'finger' not in j)
    return inb, arm < 0.15

print(f'CSV: {CSV_PATH.name}  |  {len(episodes)} episodes  |  {len(JOINTS)} joints  |  fruits={FRUITS}')
for ep in episodes:
    inb, home = ep_status(ep)
    n = len(df[df.episode == ep])
    print(f"  ep{ep}: {n} frames  |  入籃={'YES' if inb else 'no'}  回home={'YES' if home else 'no'}")
```

```python
                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                        def view(episode, frame):
    e = df[df.episode == episode].reset_index(drop=True)
    frame = int(min(frame, len(e) - 1))
    r = e.iloc[frame]
    inb, home = ep_status(episode)
    b = BASK[TARGET_BASKET]
    fig, ax = plt.subplots(1, 3, figsize=(18, 5.5))

    # (a) top view x-y
    for links, c, lab in [(RIGHT, 'tab:blue', 'right'), (LEFT, 'tab:orange', 'left')]:
        ax[0].plot([r[f'{x}_x'] for x in links], [r[f'{x}_y'] for x in links], '-o', color=c, ms=4, label=f'{lab} arm')
    for f in FRUITS:
        ax[0].scatter([r[f'{f}_x']], [r[f'{f}_y']], s=110, marker='*', label=f, zorder=5)
    ax[0].add_patch(Rectangle((RB['x'][0], RB['y'][0]), RB['x'][1]-RB['x'][0], RB['y'][1]-RB['y'][0], fill=False, ec='red'))
    ax[0].add_patch(Rectangle((YB['x'][0], YB['y'][0]), YB['x'][1]-YB['x'][0], YB['y'][1]-YB['y'][0], fill=False, ec='goldenrod'))
    ax[0].scatter([0], [0], marker='s', c='k', s=30, label='home~(0,0)')
    ax[0].set(xlabel='x (m)', ylabel='y (m)', title='TOP view (x-y): arms + fruits + baskets', xlim=(-0.1, 0.7), ylim=(-0.4, 0.4))
    ax[0].legend(fontsize=7, loc='upper right'); ax[0].grid(alpha=.3); ax[0].set_aspect('equal')

    # (b) side view x-z (platform collision check)
    for links, c in [(RIGHT, 'tab:blue'), (LEFT, 'tab:orange')]:
        ax[1].plot([r[f'{x}_x'] for x in links], [r[f'{x}_z'] for x in links], '-o', color=c, ms=4)
    for f in FRUITS:
        ax[1].scatter([r[f'{f}_x']], [r[f'{f}_z']], s=110, marker='*', zorder=5)
    ax[1].add_patch(Rectangle((PLAT['x'][0], 0), PLAT['x'][1]-PLAT['x'][0], PLAT['top'], color='brown', alpha=.18))
    ax[1].axhline(PLAT['top'], ls=':', c='brown', label='platform top 0.193')
    ax[1].set(xlabel='x (m)', ylabel='z (m)', title='SIDE view (x-z): platform-hit check', xlim=(-0.1, 0.7), ylim=(0, 0.65))
    ax[1].legend(fontsize=7); ax[1].grid(alpha=.3)

    # (c) 16 joints
    names = [j.replace('openarm_', '').replace('_joint', '_j').replace('_finger_j1', '_grip') for j in JOINTS]
    colors = ['tab:blue' if 'right' in j else 'tab:orange' for j in JOINTS]
    ax[2].barh(range(len(JOINTS)), [r[j] for j in JOINTS], color=colors)
    ax[2].set_yticks(range(len(JOINTS))); ax[2].set_yticklabels(names, fontsize=7)
    ax[2].axvline(0, c='k', lw=0.5); ax[2].invert_yaxis(); ax[2].grid(alpha=.3, axis='x')
    ax[2].set(xlabel='rad', title='16 joints (blue=right, orange=left)', xlim=(-2.2, 2.2))

    fig.suptitle(f"ep{episode}  frame {frame}/{len(e)-1}   |   in-basket={'YES' if inb else 'NO'}   arm-home={'YES' if home else 'NO'}",
                 fontsize=14, fontweight='bold')
    plt.tight_layout(); plt.show()

ep_dd = widgets.Dropdown(options=episodes, description='episode')
fr_sl = widgets.IntSlider(min=0, max=len(df[df.episode == episodes[0]]) - 1, step=1, value=0,
                          description='frame', continuous_update=False, layout=widgets.Layout(width='70%'))
def _on_ep(ch):
    fr_sl.max = len(df[df.episode == ch['new']]) - 1; fr_sl.value = 0
ep_dd.observe(_on_ep, names='value')
widgets.interactive(view, episode=ep_dd, frame=fr_sl)
```
